# Hospital Operations Data Analysis: Patient Visits and Service Performance

**Project type:** Educational exploratory data analysis (EDA)  
**Dataset:** `Hospital_Operations_Dataset.csv` (500 synthetic hospital visit records)

## Project prompt
You are a junior data analyst working with a fictional hospital. The management team wants to understand patient flow and service performance. Analyse how department, visit type, waiting time, consultation time, length of stay, billing, admission status and satisfaction are connected.

**Tasks**
1. Load and inspect the hospital dataset.
2. Check data types, missing values and duplicate records.
3. Clean missing numeric values using median imputation.
4. Calculate descriptive statistics and department-wise averages.
5. Create visualizations: department visits, waiting-time distribution, average waiting time, average billing, admission status and satisfaction by visit type.
6. Summarize findings and limitations in simple language.

> **Important:** This is a fully synthetic educational dataset. It does not contain real patient records and must not be used to evaluate a real hospital or make medical decisions.


```python
# 1. Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline
```

```python
# 2. Load the dataset
# Keep Hospital_Operations_Dataset.csv in the same folder as this notebook.
df = pd.read_csv("Hospital_Operations_Dataset.csv")
df.head()
```

```python
# 3. Basic dataset information
print("Shape:", df.shape)
print("\\nColumn information:")
df.info()
```

```python
# 4. Check missing values and duplicate rows
print("Missing values by column:")
print(df.isnull().sum())

print("\\nDuplicate rows:", df.duplicated().sum())
```

```python
# 5. Clean the data
clean_df = df.copy()

numeric_columns = [
    "Wait_Time_Minutes",
    "Satisfaction_Score",
    "Billing_Amount_INR"
]

for column in numeric_columns:
    clean_df[column] = clean_df[column].fillna(clean_df[column].median())

print("Missing values after cleaning:")
print(clean_df.isnull().sum())
```

```python
# 6. Descriptive statistics
clean_df.describe(include="all").round(2)
```

```python
# 7. Patient visits by department
print(clean_df["Department"].value_counts())

clean_df["Department"].value_counts().plot(kind="bar",
    title="Patient Visits by Department")
plt.xlabel("Department")
plt.ylabel("Number of Visits")
plt.xticks(rotation=25, ha="right")
plt.tight_layout()
plt.show()
```

```python
# 8. Waiting-time distribution
plt.figure(figsize=(7, 4))
plt.hist(clean_df["Wait_Time_Minutes"], bins=15, edgecolor="black")
plt.title("Distribution of Patient Waiting Time")
plt.xlabel("Waiting Time (minutes)")
plt.ylabel("Number of Patients")
plt.show()
```

```python
# 9. Average waiting time by department
clean_df.groupby("Department")["Wait_Time_Minutes"].mean().sort_values().plot(
    kind="bar", title="Average Waiting Time by Department")
plt.xlabel("Department")
plt.ylabel("Average Waiting Time (minutes)")
plt.xticks(rotation=25, ha="right")
plt.tight_layout()
plt.show()
```

```python
# 10. Average billing amount by department
clean_df.groupby("Department")["Billing_Amount_INR"].mean().sort_values().plot(
    kind="bar", title="Average Billing Amount by Department")
plt.xlabel("Department")
plt.ylabel("Average Bill (INR)")
plt.xticks(rotation=25, ha="right")
plt.tight_layout()
plt.show()
```

```python
# 11. Admission status distribution
print(clean_df["Admission_Status"].value_counts())

clean_df["Admission_Status"].value_counts().plot(
    kind="bar", title="Admission Status Distribution")
plt.xlabel("Admission Status")
plt.ylabel("Number of Patients")
plt.xticks(rotation=0)
plt.show()
```

```python
# 12. Average satisfaction by visit type
clean_df.groupby("Visit_Type")["Satisfaction_Score"].mean().plot(
    kind="bar", title="Average Satisfaction by Visit Type")
plt.xlabel("Visit Type")
plt.ylabel("Average Satisfaction (1–5)")
plt.xticks(rotation=0)
plt.show()
```

```python
# 13. Department-wise performance summary
department_summary = clean_df.groupby("Department").agg(
    Patient_Visits=("Patient_ID", "count"),
    Average_Wait_Minutes=("Wait_Time_Minutes", "mean"),
    Average_Consultation_Minutes=("Consultation_Minutes", "mean"),
    Average_Bill_INR=("Billing_Amount_INR", "mean"),
    Average_Satisfaction=("Satisfaction_Score", "mean")
).round(2)

department_summary
```

```python
# 14. Key summary
total_records = len(clean_df)
admitted_records = (clean_df["Admission_Status"] == "Admitted").sum()

print("Total patient visits:", total_records)
print("Admitted patients:", admitted_records)
print("Admission share:", round(admitted_records / total_records * 100, 2), "%")
print("Average waiting time:", round(clean_df["Wait_Time_Minutes"].mean(), 2), "minutes")
print("Average satisfaction:", round(clean_df["Satisfaction_Score"].mean(), 2), "out of 5")
print("\\nNote: All records are synthetic and for classroom practice only.")
```
